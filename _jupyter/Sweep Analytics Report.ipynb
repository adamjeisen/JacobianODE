```python
import torch
import numpy as np


def compute_lyapunov_vectors(jacs: torch.Tensor, dt: float) -> tuple[torch.Tensor, torch.Tensor]:
    """Compute Gram-Schmidt orthonormal Lyapunov vectors via QR decomposition.

    Performs the same forward QR iteration as the Lyapunov exponent computation,
    but returns the orthonormal Q matrices at every timestep.

    Parameters
    ----------
    jacs : torch.Tensor
        Jacobian matrices along trajectories, shape ``(B, T, D, D)`` or ``(T, D, D)``.
    dt : float
        Time step between successive Jacobians.

    Returns
    -------
    Q_all : torch.Tensor
        Orthonormal Lyapunov vectors at each timestep, shape ``(B, T, D, D)``
        (or ``(T, D, D)`` if input was unbatched).
        ``Q_all[..., t, :, i]`` is the i-th Lyapunov vector at time t.
    exponents : torch.Tensor
        Running-average Lyapunov exponents at each timestep, shape ``(B, T, D)``
        (or ``(T, D)``). The final slice ``exponents[..., -1, :]`` gives the
        converged exponents (equivalent to ``compute_lyapunov_exponents``).
    """
    unbatched = jacs.ndim == 3
    if unbatched:
        jacs = jacs.unsqueeze(0)

    B, T, D, _ = jacs.shape
    Q = torch.eye(D, dtype=jacs.dtype, device=jacs.device).expand(B, -1, -1).clone()
    log_diag_sum = torch.zeros(B, D, dtype=jacs.dtype, device=jacs.device)

    Q_all = torch.empty(B, T, D, D, dtype=jacs.dtype, device=jacs.device)
    exponents_all = torch.empty(B, T, D, dtype=jacs.dtype, device=jacs.device)

    for t in range(T):
        M = torch.linalg.matrix_exp(jacs[:, t] * dt)
        Z = M @ Q
        Q, R = torch.linalg.qr(Z)
        # Sign correction for QR ambiguity (positive diagonal of R)
        diag_R = torch.diagonal(R, dim1=-2, dim2=-1)
        signs = torch.sign(diag_R)
        signs[signs == 0] = 1.0
        Q = Q * signs.unsqueeze(-2)
        log_diag_sum += torch.log(torch.abs(diag_R))

        Q_all[:, t] = Q
        exponents_all[:, t] = log_diag_sum / ((t + 1) * dt)

    # Sort exponents descending and reorder vectors to match
    sort_idx = exponents_all[:, -1].argsort(dim=-1, descending=True)
    # Reorder all timesteps consistently using the final-time ordering
    Q_all = Q_all.gather(-1, sort_idx[:, None, None, :].expand_as(Q_all))
    exponents_all = exponents_all.gather(-1, sort_idx[:, None, :].expand_as(exponents_all))

    if unbatched:
        Q_all = Q_all.squeeze(0)
        exponents_all = exponents_all.squeeze(0)
    return Q_all, exponents_all


# --- Encode test trajectories and compute Jacobians ---
n_target_dims = lit_model.n_target_dims

with torch.no_grad():
    traj_seq = torch.as_tensor(trajs["test_trajs"].sequence).float().to(device)
    print(f"Test trajectories shape: {traj_seq.shape}")  # (N_traj, T, D_obs)

    # Encode
    z_seq = lit_model.encode_trajectory(traj_seq)
    z_dyn = z_seq[..., :n_target_dims] if n_target_dims is not None else z_seq
    print(f"Latent dynamic shape:    {z_dyn.shape}")  # (N_traj, T', D_dyn)

    # Compute Jacobians at every point (one trajectory at a time to save memory)
    N_traj = z_dyn.shape[0]
    Q_list, exp_list = [], []
    for i in range(N_traj):
        jacs_i = lit_model.compute_jacobians(z_dyn[i:i+1])  # (1, T', D, D)
        Q_i, exp_i = compute_lyapunov_vectors(jacs_i, dt)    # (1, T', D, D), (1, T', D)
        Q_list.append(Q_i.cpu())
        exp_list.append(exp_i.cpu())
        le_final = exp_i[0, -1].cpu().numpy()
        print(f"  Traj {i}: λ = {le_final}")

    lyapunov_vectors = torch.cat(Q_list, dim=0)    # (N_traj, T', D_dyn, D_dyn)
    running_exponents = torch.cat(exp_list, dim=0)  # (N_traj, T', D_dyn)

print(f"\nLyapunov vectors shape:    {lyapunov_vectors.shape}")
print(f"Running exponents shape:   {running_exponents.shape}")
print(f"Final exponents (traj 0):  {running_exponents[0, -1].numpy()}")
```

```python
# Project dz onto Lyapunov vectors at each point
# z_dyn: (3, 1101, 7), lyapunov_vectors: (3, 1101, 7, 7)

# Compute dz = z_{t+1} - z_t, shape (3, 1100, 7)
dz = z_dyn[:, 1:, :] - z_dyn[:, :-1, :]

# Trim Lyapunov vectors to match (use vectors at time t, not t+1)
Q_t = lyapunov_vectors[:, :-1, :, :]  # (3, 1100, 7, 7)

# Project: for each point, dot dz with each Lyapunov vector column
# dz: (3, 1100, 7) -> (3, 1100, 7, 1)
# Q_t: (3, 1100, 7, 7)
# projection[..., i] = dz · Q_t[:, :, :, i]
projections = torch.einsum("btd, btdi -> bti", dz.cpu(), Q_t)  # (3, 1100, 7)

print(f"dz shape:          {dz.shape}")
print(f"Q_t shape:         {Q_t.shape}")
print(f"projections shape: {projections.shape}")
print(f"\nTraj 0, t=0 projections: {projections[0, 0].numpy()}")
print(f"Traj 0, t=0 |dz|:       {dz[0, 0].cpu().norm().item():.6f}")
print(f"Traj 0, t=0 sum(proj²): {(projections[0, 0] ** 2).sum().item():.6f}  (should match |dz|²)")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

n_target = n_target_dims
D_obs = traj_seq.shape[-1]
T = z_dyn.shape[1]

# ================================================================
# 1. Encoder Jacobian ∂z_dyn/∂x at every point
# ================================================================
print("Computing encoder Jacobians ...")

enc_jacs = []  # will be (B, T, n_target, D_obs)
for b in range(N_traj):
    rows = []
    for j in range(n_target):
        x = traj_seq[b:b+1].to(device).detach().requires_grad_(True)
        z_out = lit_model.encode_trajectory(x)
        z_out[0, :T, j].sum().backward()
        rows.append(x.grad[0, :T].cpu())  # (T, D_obs)
    J_b = torch.stack(rows, dim=1)  # (T, n_target, D_obs)
    enc_jacs.append(J_b)
    print(f"  Traj {b}: done")

enc_jacs = torch.stack(enc_jacs)  # (B, T, n_target, D_obs)
print(f"Encoder Jacobians shape: {enc_jacs.shape}")

# ================================================================
# 2. SVD → tangent basis U(t) and singular values σ(t)
# ================================================================
print("\nComputing SVD ...")
U_enc, S_enc, _ = torch.linalg.svd(enc_jacs, full_matrices=False)
# U_enc: (B, T, n_target, n_target) — orthonormal basis in R^n_target
# S_enc: (B, T, n_target)

print("Singular values (mean ± std across all points):")
for i in range(n_target):
    print(f"  σ_{i+1} = {S_enc[:,:,i].mean():.4f} ± {S_enc[:,:,i].std():.4f}")

# ================================================================
# 3. Temporal alignment via Procrustes (handles swaps + sign flips)
# ================================================================
# At each timestep, find the orthogonal rotation that best aligns U(t) to U(t-1).
# This resolves both sign flips AND column swaps from near-degenerate singular values.
print("\nAligning singular vectors via Procrustes ...")
for b in range(N_traj):
    for t in range(1, T):
        # Find orthogonal matrix R that minimizes ||U(t) R - U(t-1)||_F
        # Solution: R = V @ W^T where U(t-1)^T @ U(t) = V S W^T
        M = U_enc[b, t-1].T @ U_enc[b, t]  # (n_target, n_target)
        Vp, _, Wp = torch.linalg.svd(M)
        R = Wp.T @ Vp.T  # closest orthogonal alignment
        U_enc[b, t] = U_enc[b, t] @ R

# ================================================================
# 4. Singular value spectrum visualization
# ================================================================
S_mean = S_enc.mean(dim=(0, 1))
S_std = S_enc.std(dim=(0, 1))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

ax = axes[0]
ax.bar(range(1, n_target + 1), S_mean.numpy(), yerr=S_std.numpy(), capsize=4)
ax.set_xlabel("Index")
ax.set_ylabel("σ (log scale)")
ax.set_title("Mean encoder Jacobian singular values")
ax.set_yscale("log")

ax = axes[1]
for i in range(n_target):
    ax.plot(S_enc[0, :, i].numpy(), label=f"σ_{i+1}", alpha=0.8)
ax.set_xlabel("Time step")
ax.set_ylabel("σ (log scale)")
ax.set_title("Singular values along trajectory 0")
ax.legend(fontsize=8)
ax.set_yscale("log")

var_explained = S_mean ** 2 / (S_mean ** 2).sum()
cum_var = torch.cumsum(var_explained, dim=0)
ax = axes[2]
ax.bar(range(1, n_target + 1), cum_var.numpy())
ax.axhline(y=0.99, color='r', linestyle='--', label='99%')
ax.set_xlabel("Number of dimensions")
ax.set_ylabel("Cumulative variance explained")
ax.set_title("Intrinsic dimensionality (encoder Jacobian)")
ax.legend()

plt.tight_layout()
plt.show()

# ================================================================
# 5. Soft projector: w_i(t) = σ_i²(t) / (σ_i²(t) + ε²)
# ================================================================
# Use a user-tunable ε. Default: geometric mean of the smallest on-manifold
# and largest off-manifold singular value. Print a sweep so the user can choose.
log_S_mean = torch.log(S_mean)
gaps = log_S_mean[:-1] - log_S_mean[1:]

print("\nSingular value gaps (in log space):")
for i in range(len(gaps)):
    eps_i = torch.sqrt(S_mean[i] * S_mean[i + 1]).item()
    print(f"  gap {i+1}→{i+2}: {gaps[i]:.4f}  (ε = {eps_i:.4f})")

# Default: pick the gap that gives the most reasonable split
# Show weights for a few candidate ε values
print("\nSoft weight profiles for different ε values:")
print(f"  {'ε':>8s}", end="")
for i in range(n_target):
    print(f"  {'w_'+str(i+1):>8s}", end="")
print()
for k in range(n_target - 1):
    eps_k = torch.sqrt(S_mean[k] * S_mean[k + 1]).item()
    w_k = S_mean ** 2 / (S_mean ** 2 + eps_k ** 2)
    print(f"  {eps_k:8.4f}", end="")
    for i in range(n_target):
        print(f"  {w_k[i]:8.4f}", end="")
    print(f"  ← gap {k+1}→{k+2}")

# Default ε: use gap between σ_3 and σ_4 if it exists (expected for Lorenz-like 3D)
# Otherwise fall back to largest gap
gap_idx = min(2, len(gaps) - 1)  # prefer gap after dim 3
eps = torch.sqrt(S_mean[gap_idx] * S_mean[gap_idx + 1]).item()
print(f"\nUsing ε = {eps:.6f} (gap {gap_idx+1}→{gap_idx+2})")

W = S_enc ** 2 / (S_enc ** 2 + eps ** 2)  # (B, T, n_target)
W_sqrt = W.sqrt()
print(f"Mean soft weights: {W.mean(dim=(0,1)).numpy()}")

# ================================================================
# 6. U̇ via central finite differencing
# ================================================================
U_dot = torch.zeros_like(U_enc)
U_dot[:, 1:-1] = (U_enc[:, 2:] - U_enc[:, :-2]) / (2 * dt)
U_dot[:, 0] = (U_enc[:, 1] - U_enc[:, 0]) / dt
U_dot[:, -1] = (U_enc[:, -1] - U_enc[:, -2]) / dt

# Verify skew-symmetry of U̇ᵀU at several points
skew_errs = []
for t_check in [100, 300, 500, 700, 900]:
    geom_sample = torch.einsum("btji,btjk->btik", U_dot[:, t_check:t_check+1], U_enc[:, t_check:t_check+1])
    skew_err = (geom_sample + geom_sample.transpose(-2, -1)).abs().max().item()
    skew_errs.append(skew_err)
print(f"\nU̇ᵀU skew-symmetry check (max |U̇ᵀU + UᵀU̇|) at t=100,300,500,700,900:")
print(f"  {skew_errs}")
print(f"  Mean: {np.mean(skew_errs):.6f}")

# ================================================================
# 7. Projected Jacobian: J̄(t) = W^½ (UᵀJU + U̇ᵀU) W^½
# ================================================================
print("\nComputing projected Jacobians ...")
with torch.no_grad():
    jacs_list = []
    for b in range(N_traj):
        J_b = lit_model.compute_jacobians(z_dyn[b:b+1]).cpu()[0]
        jacs_list.append(J_b)
    jacs_dyn = torch.stack(jacs_list)  # (B, T, n_target, n_target)

# UᵀJU — dynamics Jacobian rotated into encoder SVD basis
J_rotated = torch.einsum("btji,btjk,btkl->btil", U_enc, jacs_dyn, U_enc)

# U̇ᵀU — geometric correction (connection/Christoffel term, skew-symmetric)
geom_correction = torch.einsum("btji,btjk->btik", U_dot, U_enc)

# Full tangent Jacobian (exact, no information lost — just a basis change + correction)
J_tangent = J_rotated + geom_correction

# Soft-projected Jacobian (discounts off-manifold directions)
J_projected = J_tangent * W_sqrt.unsqueeze(-1) * W_sqrt.unsqueeze(-2)

print(f"\nResults:")
print(f"  U_enc         {U_enc.shape}        — encoder SVD basis at each point")
print(f"  S_enc         {S_enc.shape}     — singular values (manifold dimension spectrum)")
print(f"  W             {W.shape}     — soft weights per direction")
print(f"  J_tangent     {J_tangent.shape}  — full tangent Jacobian (UᵀJU + U̇ᵀU)")
print(f"  J_projected   {J_projected.shape}  — soft-projected Jacobian")
print(f"  geom_correction {geom_correction.shape} — U̇ᵀU connection term")
```

```python
# Lyapunov exponents from the soft-projected Jacobian
from JacobianODE.models.latent_jacobian import LitLatentJacobianODE

print("Computing Lyapunov exponents from J_projected ...")
lyap_projected = LitLatentJacobianODE.compute_lyapunov_exponents(J_projected, dt)
# (B, n_target) — one spectrum per trajectory

print("\nSoft-projected Lyapunov exponents per trajectory:")
for b in range(N_traj):
    le = lyap_projected[b].numpy()
    print(f"  Traj {b}: {le}")

lyap_proj_mean = lyap_projected.mean(dim=0)
lyap_proj_std = lyap_projected.std(dim=0)
print(f"\nMean ± std:")
for i in range(n_target):
    print(f"  λ_{i+1} = {lyap_proj_mean[i]:+.4f} ± {lyap_proj_std[i]:.4f}")

# Compare with unweighted tangent Jacobian and raw dynamics Jacobian
lyap_tangent = LitLatentJacobianODE.compute_lyapunov_exponents(J_tangent, dt)
lyap_raw = LitLatentJacobianODE.compute_lyapunov_exponents(jacs_dyn, dt)

print("\n--- Comparison ---")
print(f"{'':>24s} {'Projected':>12s} {'Tangent (no W)':>14s} {'Raw dynamics':>14s}" +
      (f" {'True':>10s}" if TRUE_LYAPUNOV else ""))
for i in range(n_target):
    row = f"  λ_{i+1}:                "
    row += f"{lyap_proj_mean[i]:+10.4f}  "
    row += f"{lyap_tangent.mean(0)[i]:+12.4f}  "
    row += f"{lyap_raw.mean(0)[i]:+12.4f}  "
    if TRUE_LYAPUNOV and i < len(TRUE_LYAPUNOV):
        row += f"{TRUE_LYAPUNOV[i]:+8.4f}"
    print(row)
```

```python
from JacobianODE.models.latent_jacobian import LitLatentJacobianODE

# ================================================================
# 1. Project dz onto encoder SVD basis at each point
# ================================================================
# dz: (B, T-1, n_target), U_enc: (B, T, n_target, n_target)
dz_latent = z_dyn[:, 1:, :] - z_dyn[:, :-1, :]  # (B, T-1, n_target)
U_t = U_enc[:, :-1]  # use U at time t, not t+1

# Project dz onto each SVD direction: proj_i = dz · u_i
dz_proj = torch.einsum("btd, btdi -> bti", dz_latent.cpu(), U_t)  # (B, T-1, n_target)

# Variance of projections per direction (across all points and trajectories)
var_per_dir = dz_proj.var(dim=(0, 1))  # (n_target,)
frac_var = var_per_dir / var_per_dir.sum()

print("dz variance in encoder SVD basis:")
for i in range(n_target):
    print(f"  dir {i+1}: var = {var_per_dir[i]:.6f}  ({frac_var[i]*100:.2f}%)")
print(f"  cumulative: {torch.cumsum(frac_var, dim=0).numpy()}")

# ================================================================
# 2. Construct weights: w_i = Var(proj_i) / max_j Var(proj_j)
# ================================================================
# Normalize so max weight = 1 (base case W=I is recovered when all vars are equal)
# Use per-point variance? Or global? Global is cleaner — the manifold structure
# shouldn't change point-to-point, just the direction of the tangent vectors.
W_dz = var_per_dir / var_per_dir.max()  # (n_target,) — global weights, max = 1
W_dz_sqrt = W_dz.sqrt()

print(f"\nSoft weights (max-normalized dz variance): {W_dz.numpy()}")
print(f"Sum of weights: {W_dz.sum():.4f}  (base case = {n_target})")

# ================================================================
# 3. Apply to tangent Jacobian: J̄ = W^½ (UᵀJU + U̇ᵀU) W^½
# ================================================================
# J_tangent already computed in previous cell: (B, T, n_target, n_target)
# W_dz_sqrt is global (same at every point), so broadcast
J_projected_dz = J_tangent * W_dz_sqrt.unsqueeze(-1) * W_dz_sqrt.unsqueeze(-2)

# ================================================================
# 4. Lyapunov exponents comparison
# ================================================================
print("\nComputing Lyapunov exponents ...")
lyap_proj_dz = LitLatentJacobianODE.compute_lyapunov_exponents(J_projected_dz, dt)
lyap_tangent = LitLatentJacobianODE.compute_lyapunov_exponents(J_tangent, dt)
lyap_raw = LitLatentJacobianODE.compute_lyapunov_exponents(jacs_dyn, dt)

print(f"\n{'':>6s} {'dz-weighted':>12s} {'Tangent (W=I)':>14s} {'Raw dynamics':>14s}" +
      (f" {'True':>10s}" if TRUE_LYAPUNOV else ""))
for i in range(n_target):
    row = f"  λ_{i+1}: "
    row += f"{lyap_proj_dz.mean(0)[i]:+10.4f}  "
    row += f"{lyap_tangent.mean(0)[i]:+12.4f}  "
    row += f"{lyap_raw.mean(0)[i]:+12.4f}  "
    if TRUE_LYAPUNOV and i < len(TRUE_LYAPUNOV):
        row += f"{TRUE_LYAPUNOV[i]:+8.4f}"
    print(row)

# Also show per-trajectory for the dz-weighted version
print(f"\ndz-weighted Lyapunov exponents per trajectory:")
for b in range(N_traj):
    print(f"  Traj {b}: {lyap_proj_dz[b].numpy()}")
```

```python
from JacobianODE.models.latent_jacobian import LitLatentJacobianODE
import matplotlib.pyplot as plt

# ================================================================
# Hard projection sweep: J̄_d = U_d^T J U_d + U̇_d^T U_d for d = 1..n_target
# ================================================================
# U_enc: (B, T, n_target, n_target), columns ordered by encoder SVD singular value
# jacs_dyn: (B, T, n_target, n_target), dynamics Jacobian in latent space
# U_dot: (B, T, n_target, n_target), temporal derivative of U

print("Computing tangent-space Lyapunov exponents for d = 1..{} ...\n".format(n_target))

lyap_by_d = {}  # d -> (B, d) tensor of exponents
for d in range(1, n_target + 1):
    U_d = U_enc[..., :d]          # (B, T, n_target, d)
    Ud_d = U_dot[..., :d]         # (B, T, n_target, d)

    # U_d^T J U_d: (B, T, d, d)
    J_proj = torch.einsum("btji,btjk,btkl->btil", U_d, jacs_dyn, U_d)

    # U̇_d^T U_d: geometric correction (B, T, d, d)
    geom = torch.einsum("btji,btjk->btik", Ud_d, U_d)

    J_bar = J_proj + geom
    lyap_d = LitLatentJacobianODE.compute_lyapunov_exponents(J_bar, dt)  # (B, d)
    lyap_by_d[d] = lyap_d

    mean = lyap_d.mean(dim=0).numpy()
    std = lyap_d.std(dim=0).numpy()
    exps_str = "  ".join(f"{m:+.4f}±{s:.4f}" for m, s in zip(mean, std))
    print(f"  d={d}: {exps_str}")

# Print comparison table
print(f"\n{'d':>3s}", end="")
for i in range(n_target):
    print(f"  {'λ_'+str(i+1):>12s}", end="")
print(f"  {'True':>10s}" if TRUE_LYAPUNOV else "")

for d in range(1, n_target + 1):
    row = f"  {d}"
    mean = lyap_by_d[d].mean(dim=0)
    for i in range(d):
        row += f"  {mean[i]:+10.4f}  "
    # Pad remaining columns
    for i in range(d, n_target):
        row += f"  {'—':>10s}  "
    if TRUE_LYAPUNOV:
        row += "  |"
        for i in range(min(d, len(TRUE_LYAPUNOV))):
            row += f" {TRUE_LYAPUNOV[i]:+.4f}"
    print(row)

# ================================================================
# Visualization: Lyapunov exponents as a function of d
# ================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: all exponents across d
ax = axes[0]
# Track how each exponent index evolves as d increases
for i in range(n_target):
    ds = list(range(i + 1, n_target + 1))
    means = [lyap_by_d[d].mean(0)[i].item() for d in ds]
    stds = [lyap_by_d[d].std(0)[i].item() for d in ds]
    ax.errorbar(ds, means, yerr=stds, marker='o', markersize=4, capsize=3, label=f"λ_{i+1}")

if TRUE_LYAPUNOV:
    for i, le in enumerate(TRUE_LYAPUNOV):
        ax.axhline(y=le, color=f"C{i}", linestyle='--', alpha=0.5)

ax.set_xlabel("Projection dimension d")
ax.set_ylabel("Lyapunov exponent")
ax.set_title("Tangent-space Lyapunov spectrum vs projection dimension")
ax.legend(fontsize=8)
ax.set_xticks(range(1, n_target + 1))
ax.axhline(y=0, color='k', linestyle='-', alpha=0.2)

# Right: just the top 3 exponents zoomed in (for comparison with true Lorenz)
ax = axes[1]
for i in range(min(3, n_target)):
    ds = list(range(i + 1, n_target + 1))
    means = [lyap_by_d[d].mean(0)[i].item() for d in ds]
    stds = [lyap_by_d[d].std(0)[i].item() for d in ds]
    ax.errorbar(ds, means, yerr=stds, marker='o', markersize=5, capsize=3, label=f"λ_{i+1}")

if TRUE_LYAPUNOV:
    for i in range(min(3, len(TRUE_LYAPUNOV))):
        ax.axhline(y=TRUE_LYAPUNOV[i], color=f"C{i}", linestyle='--', alpha=0.5,
                    label=f"true λ_{i+1} = {TRUE_LYAPUNOV[i]}")

ax.set_xlabel("Projection dimension d")
ax.set_ylabel("Lyapunov exponent")
ax.set_title("Top 3 exponents vs d (dashed = true Lorenz)")
ax.legend(fontsize=8)
ax.set_xticks(range(1, n_target + 1))
ax.axhline(y=0, color='k', linestyle='-', alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
from JacobianODE.models.latent_jacobian import LitLatentJacobianODE
import matplotlib.pyplot as plt
from scipy.linalg import schur as scipy_schur, rsf2csf

# ================================================================
# Approach 1: Raw SVD projection (no Procrustes, no U̇ correction)
# ================================================================
# Recompute SVD fresh (un-Procrusted) — columns ordered by σ at each point
print("Recomputing raw SVD (no Procrustes) ...")
U_raw, S_raw, _ = torch.linalg.svd(enc_jacs, full_matrices=False)
# U_raw: (B, T, n_target, n_target) — columns ordered by σ_1 > σ_2 > ... at each t

print("\n=== Approach 1: Raw SVD projection (U_d^T J U_d, no geometric correction) ===\n")
lyap_svd_proj = {}
for d in range(1, n_target + 1):
    U_d = U_raw[..., :d]  # (B, T, n_target, d) — true top-d subspace
    J_proj = torch.einsum("btji,btjk,btkl->btil", U_d, jacs_dyn, U_d)  # (B, T, d, d)
    lyap_d = LitLatentJacobianODE.compute_lyapunov_exponents(J_proj, dt)
    lyap_svd_proj[d] = lyap_d
    mean = lyap_d.mean(dim=0).numpy()
    std = lyap_d.std(dim=0).numpy()
    exps_str = "  ".join(f"{m:+.4f}±{s:.4f}" for m, s in zip(mean, std))
    print(f"  d={d}: {exps_str}")

# ================================================================
# Approach 2: Schur decomposition of dynamics Jacobian
# ================================================================
# At each point, Schur decompose J = Q T Q^T, order eigenvalues by
# descending real part. Top-left d×d block of T = on-manifold dynamics.
print("\n=== Approach 2: Schur decomposition of J(t) ===\n")

# Compute ordered real Schur form at each point
B_n, T_n = jacs_dyn.shape[:2]
J_schur_blocks = {}  # d -> (B, T, d, d)

# Compute Schur for all points, extract blocks for each d
# Use scipy for ordered Schur decomposition
jacs_np = jacs_dyn.numpy()
schur_T = np.empty_like(jacs_np)  # (B, T, n, n) upper triangular Schur form

print("  Computing ordered Schur decompositions ...")
for b in range(B_n):
    for t in range(T_n):
        J_pt = jacs_np[b, t]
        # Real Schur form, sorted so eigenvalues with largest real part come first
        T_s, Q_s = scipy_schur(J_pt, output='real', sort=lambda r, i: -r)
        schur_T[b, t] = T_s

schur_T_torch = torch.from_numpy(schur_T).float()

lyap_schur = {}
for d in range(1, n_target + 1):
    T_block = schur_T_torch[:, :, :d, :d]  # (B, T, d, d) — top-left block
    lyap_d = LitLatentJacobianODE.compute_lyapunov_exponents(T_block, dt)
    lyap_schur[d] = lyap_d
    mean = lyap_d.mean(dim=0).numpy()
    std = lyap_d.std(dim=0).numpy()
    exps_str = "  ".join(f"{m:+.4f}±{s:.4f}" for m, s in zip(mean, std))
    print(f"  d={d}: {exps_str}")

# ================================================================
# Comparison table
# ================================================================
print(f"\n{'d':>3s}  {'--- SVD projection ---':>40s}  {'--- Schur block ---':>40s}" +
      ("  True" if TRUE_LYAPUNOV else ""))
for d in range(1, n_target + 1):
    svd_mean = lyap_svd_proj[d].mean(0)
    sch_mean = lyap_schur[d].mean(0)
    svd_str = "  ".join(f"{svd_mean[i]:+.4f}" for i in range(d))
    sch_str = "  ".join(f"{sch_mean[i]:+.4f}" for i in range(d))
    true_str = ""
    if TRUE_LYAPUNOV:
        true_str = "  | " + "  ".join(f"{TRUE_LYAPUNOV[i]:+.4f}" for i in range(min(d, len(TRUE_LYAPUNOV))))
    print(f"  {d}  {svd_str:>40s}  {sch_str:>40s}{true_str}")

# ================================================================
# Visualization
# ================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, data, title in [
    (axes[0], lyap_svd_proj, "SVD projection (U_d^T J U_d)"),
    (axes[1], lyap_schur, "Schur block (top-left d×d of T)"),
]:
    for i in range(n_target):
        ds = list(range(i + 1, n_target + 1))
        means = [data[d].mean(0)[i].item() for d in ds]
        stds = [data[d].std(0)[i].item() for d in ds]
        ax.errorbar(ds, means, yerr=stds, marker='o', markersize=4, capsize=3, label=f"λ_{i+1}")
    if TRUE_LYAPUNOV:
        for i, le in enumerate(TRUE_LYAPUNOV):
            ax.axhline(y=le, color=f"C{i}", linestyle='--', alpha=0.4)
    ax.set_xlabel("Projection dimension d")
    ax.set_ylabel("Lyapunov exponent")
    ax.set_title(title)
    ax.legend(fontsize=7)
    ax.set_xticks(range(1, n_target + 1))
    ax.axhline(y=0, color='k', linestyle='-', alpha=0.2)

plt.tight_layout()
plt.show()
```
